#### 6.- Devuelve la secuencia de ARN que surgiría a partir de la siguiente secuencia. Recuerda que la secuencia de ARN resultante tiene que ser complementaria a la de ADN y además, hay que cambiar un nucleótido. (1 pto)

```python
#Realizamos un def para definir la función que transforme la secuencia de ADN a su cadena complementaria de ARN
def adn_a_arn(secuencia_adn):
# Creamos un diccionario para definir cada base complementaria de nuestra secuencia de ADN a ARN(A-U,T-A,G-C,C-G)
    complementaria = {"A": "U", "T": "A", "G": "C", "C": "G"}
# Creamos la secuencia de ARN, complementearia a nuestra secuencia de ADN. 
#Nuestra nueva tupla(secuencia_arn) la igualamos a la función join para que una todos las filas de nuestro archivo en una sola, 
#recorriendo nuestra secuencia de ADN pero cambiando cada base a su complementaria usando el diccionario llamado complementaria 
    secuencia_arn = "".join(complementaria[base] for base in secuencia_adn)
    return secuencia_arn
```

```python
# Comprobamos que la secuencia de ARN es la complementaria a la de ADN y la imprimimos llamandola "Secuencia de ARN complementaria:"
secuencia_arn = adn_a_arn(secuencia_adn)
print("Secuencia de ARN complementaria:",secuencia_arn)
```

#### 7.- Realiza una función que realice una digestión con las siguientes enzimas de restricción en nuestra secuencia. La función tiene que devolver la siguiente información: (2 pto)

2. Enzimas que no tienen sitios de corte (si las hay)
3. Resultado de la digestión: los fragmentos resultantes de nuestra secuencia
4. Indicar número y tamaño de cada fragmento

```python
import re
# Diccionario con las enzimas de restricción y sus sitios de corte
enzR = {
    'HindIII' : 'A*AGCTT',
    'BamHI' : 'G*GATCC',
    'AluI' : 'AG*CT',
    'Sau3AI' : '*GATC',
    'EcoRI' : 'G*AATTC'
}
```

```python
#Utilizamos la función def para definir la funcíon de digestion de secuencia
def digestion_adn(secuencia_adn, enzimas):
    # Creamos diccionarios que almacenen los resultados de cada digestión
    enzimas_cortan = {}
    enzimas_no_cortan = []
    fragmentos_resultantes = []
    
    # Función que al correr la secuencia guarde cada enzima de digestión y su sitio de corte
    for enzima, sitio_corte in enzimas.items():
        # Obtenemos el sitio de corte y la posición del asterisco
        sitio = sitio_corte.replace('*', '')
        corte_index = sitio_corte.index('*')
        
        # Buscamos en la secuencia todos los sitios de corte
        posiciones = [m.start() + corte_index for m in re.finditer(sitio, secuencia_adn)]
        
        # Añadimos cada enzima y posición de corte y es que hay sitios de corte
        if posiciones:
            enzimas_cortan[enzima] = posiciones
            # Generar los fragmentos dividiendo la secuencia en cada sitio de corte encontrado
            cortes = [0] + [pos + corte_index for pos in posiciones] + [len(secuencia_adn)]
            fragmentos = [secuencia_adn[cortes[i]:cortes[i+1]] for i in range(len(cortes) - 1)]
            fragmentos_resultantes.extend(fragmentos)
        else:
            # Si no hay sitio de corte, añadimos la enzima a enzimas_no_cortan
            enzimas_no_cortan.append(enzima)
    
    # Quitamos cada fragmento vacío y calculamos el tamaño de cada fragmento
    fragmentos_limpios = [frag for frag in fragmentos_resultantes if frag]
    info_fragmentos = [(len(frag), frag) for frag in fragmentos_limpios]
    
    # Definimos los resultados para cada apartado
    resultado = {
        "enzimas_con_corte": enzimas_cortan,
        "enzimas_sin_corte": enzimas_no_cortan,
        "fragmentos": fragmentos_limpios,
        "numero_fragmentos": len(fragmentos_limpios),
        "tamanos_fragmentos": [len(frag) for frag in fragmentos_limpios]
    }
    
    return resultado
secuencia_adn
resultado = digestion_adn(secuencia_adn, enzR)
```

```python
#Mostramos con la función print los resultados del primer apartado con las enzimas de corte y sus respectivas posiciones 
print("Enzimas con sitios de corte y posiciones:", resultado["enzimas_con_corte"])
```

```python
#Mostramos con la función print los resultados del segundo apartado con las enzimas sin sitios de corte en caso de que haya, sino solo serán nombradas
print("Enzimas sin sitios de corte:", resultado["enzimas_sin_corte"])
```

```python
#Mostramos con la función print los resultados del tercer apartado, los fragmentos resultantes de nuestra secuencia
print("Fragmentos resultantes de la digestión:", resultado["fragmentos"])
```
